#  <span style="color:green">1. What is hypothesis testing in statistics?</span>

Hypothesis testing is a statistical method used to decide whether there is enough evidence in a sample of data to infer that a certain condition holds true for the entire population.

It involves:

- Null Hypothesis (H₀): The default assumption (e.g., no difference, no effect).

- Alternative Hypothesis (H₁ or Ha): What you want to prove (e.g., there is a difference).

- Test Statistic: A value calculated from the sample to test the hypothesis.

- P-value: The probability of getting a result as extreme as the test statistic under H₀.

- Decision Rule: If p-value < α (significance level, like 0.05), reject H₀.



#  <span style="color:green">2. What is the null hypothesis, and how does it differ from the alternative hypothesis?
</span>

In hypothesis testing, we always test two competing hypotheses:

🔹 Null Hypothesis (H₀):
 - It is the default assumption.
 - States that there is no effect, no difference, or the status quo remains.
 - Example:
    H₀: μ = 100 (The population mean is 100)

🔹 Alternative Hypothesis (H₁ or Ha):
 - It is what we want to prove or support.
 - States that there is an effect, difference, or change.
 - Example:
    H₁: μ ≠ 100 (The population mean is not 100)


                                                           💡 Key Differences:
| Feature                 | Null Hypothesis (H₀)       | Alternative Hypothesis (H₁)     |
|-------------------------|-----------------------------|----------------------------------|
| Assumption              | No difference/effect        | There is a difference/effect     |
| Goal                    | Try to **reject** it        | Try to **support** it            |
| Type of test            | Default for all tests       | Leads to action if H₀ is rejected |
| P-value comparison      | Used to determine if H₀ is rejected | -                      |

#  <span style="color:green"> 3.What is the significance level in hypothesis testing, and why is it important?</span>

The significance level (denoted by α, alpha) is the threshold probability used to decide whether to reject the null hypothesis (H₀).

🔹 Typical Values:
   - Common α values: 0.05 (5%), 0.01 (1%), or 0.10 (10%)

🔹 Why is it Important?
   - It defines the maximum risk you're willing to take of making a Type I error (rejecting H₀ when it’s actually true).
   - It helps you control false positives in your statistical conclusions.

🔹 Interpretation Example:
   - If α = 0.05, you're willing to accept a 5% chance of wrongly rejecting a true null hypothesis.

#  <span style="color:green">4.What does a P-value represent in hypothesis testing?
</span>

The p-value is the probability of observing a test statistic as extreme as, or more extreme than, the one obtained from your sample data, assuming that the null hypothesis (H₀) is true.

🔹 Key Points:
  - It helps you decide whether to reject H₀.
  - A small p-value (typically < α = 0.05) → strong evidence against H₀.
  - A large p-value → weak evidence against H₀ → fail to reject H₀.

🔹 Interpretation Example:
  - If p-value = 0.03 and α = 0.05:
     -  Since 0.03 < 0.05 → Reject H₀ → result is statistically significant.
  - If p-value = 0.08:
     - Since 0.08 > 0.05 → Fail to reject H₀ → not statistically significant.



#  <span style="color:green">5.How do you interpret the P-value in hypothesis testing?
</span>

The p-value tells you how likely it is to get the observed sample results if the null hypothesis (H₀) were true.



                                 🔹 Interpretation Rules (based on a chosen significance level α):

| **P-value**         | **Interpretation**                              | **Decision**                  |
|----------------------|--------------------------------------------------|--------------------------------|
| p-value < α (e.g. 0.05) | Strong evidence against H₀                   | Reject H₀ (statistically significant) |
| p-value ≥ α           | Weak evidence against H₀ or supports H₀       | Fail to reject H₀ (not significant)   |

🔹 Example Interpretation:
 - p = 0.02, α = 0.05 → 2% chance result occurred under H₀
   → Statistically significant → Reject H₀

 - p = 0.10, α = 0.05 → 10% chance result occurred under H₀
   → Not significant → Fail to reject H₀



#  <span style="color:green">6.What are Type 1 and Type 2 errors in hypothesis testing?
</span>

In hypothesis testing, errors can occur because we make decisions based on probabilities.



🔹Type 1 Error (False Positive)
  - Definition: Rejecting the null hypothesis (H₀) when it is actually true.
  - Risk Level: Controlled by the significance level (α), usually 0.05.
  - Example: Concluding a new drug works when it actually doesn’t.

🔹 Type 2 Error (False Negative)
  - Definition: Failing to reject H₀ when it is actually false.
  - Risk Level: Denoted by β (beta); 1 - β = Power of the test.
  - Example: Concluding a new drug doesn’t work when it actually does.



                                                                📊 Summary Table

| Error Type     | Decision Made                  | Reality                        | Example                                  |
|----------------|-------------------------------|--------------------------------|------------------------------------------|
| **Type 1 (α)** | Reject H₀                     | H₀ is actually true           | False alarm (detecting an effect that isn't there) |
| **Type 2 (β)** | Fail to reject H₀             | H₀ is actually false          | Missed detection (failing to detect a real effect) |

#  <span style="color:green"> 7.What is the difference between a one-tailed and a two-tailed test in hypothesis testing?
</span>

The difference lies in what direction you are testing for an effect or difference:

🔹 One-Tailed Test
  - Tests for an effect in one specific direction (greater than or less than).
  - Used when you have a clear directional hypothesis.
      Example:
          H₀: μ ≤ 100
          H₁: μ > 100 → right-tailed test

🔹 Two-Tailed Test
  - Tests for an effect in both directions (either greater or less).
  - Used when any deviation from the null matters.
      Example:
          H₀: μ = 100
          H₁: μ ≠ 100 → two-tailed test

                                                          📊 Comparison Table

| Feature              | One-Tailed Test             | Two-Tailed Test               |
|----------------------|-----------------------------|-------------------------------|
| Direction            | One (>, or <)               | Both (> and <)                |
| Critical region      | One side of the distribution | Both sides of the distribution |
| P-value              | Area in one tail            | Area in both tails (×2)       |
| Use Case             | Directional claim           | Non-directional claim         |

#  <span style="color:green">9. How do you calculate the Z-score, and what does it represent in hypothesis testing?

</span>

A Z-score measures how many standard deviations a data point (or sample mean) is from the population mean.

🔹 Z-score Formula (Single Data Point)

       Z = (X - μ) / σ
 Where:
  -  X   = individual data point
  - μ  = population mean
  - σ = population standard deviation




🔹 Z-score Formula (Sample Mean)

        Z = (X_bar - μ) / (σ / sqrt(n))

    Where:
  - X_bar = sample mean
  -  μ    = population mean
  - σ = population standard deviation
  - n     = sample size


🔹 What Does It Represent?
 - Shows how extreme your value is relative to the population.
 - In hypothesis testing, it’s used to determine the p-value.

                                                       📊 Z-score Interpretation Table

| **Z-score** | **Interpretation (in standard normal dist.)** |
|-------------|-----------------------------------------------|
| 0           | Exactly at the mean                          |
| +1 / -1     | 1 SD above/below mean                        |
| > +2 or < -2| Far from mean (usually significant)          |

#  <span style="color:green">14. What is a confidence interval, and how is it used to interpret statistical results?

</span>

A confidence interval (CI) is a range of values derived from sample data that is likely to contain the true population parameter (such as the mean) with a specified level of confidence (e.g., 95%).



🔹 Formula for Confidence Interval (for Mean):

If population standard deviation 
σ is known:

 CI = X̄ ± Z_(α/2) * (σ / √n)

 Where:
 -  X̄      = sample mean
 -  σ      = population standard deviation
 - n      = sample size
 - Z_(α/2) = critical Z-value (e.g., 1.96 for 95% confidence)

If σ is unknown, use:

 CI = X̄ ± t_(α/2, df) * (s / √n)

 Where:
 -  X̄         = sample mean
 -  s         = sample standard deviation
 - n         = sample size
 - df        = degrees of freedom (n - 1)
 -  t_(α/2)   = critical T-value from t-distribution


 


✅ Interpretation:

A 95% confidence interval means:
"We are 95% confident that the true population mean lies within this interval."

It does not mean there's a 95% chance the parameter is in the interval — the parameter is fixed; the interval is random.

#  <span style="color:green">15. What is the margin of error, and how does it affect the confidence interval?


</span>

The margin of error (MOE) is the amount added and subtracted from a sample statistic (usually the mean) to create a confidence interval.
It reflects the maximum expected difference between the true population value and the sample estimate due to random sampling variability.



🔹 Margin of Error Formula:

If population standard deviation σ is known:

MOE = Z_(α/2) * (σ / √n)

 Where:
 -  Z_(α/2) = critical value from standard normal distribution
  - σ       = population standard deviation
 - n       = sample size

 
If 
σ is unknown (use sample standard deviation 
𝑠):

MOE = t_(α/2, df) * (s / √n)

 Where:
-  t_(α/2, df) = critical t-value from t-distribution with df = n - 1
-  s           = sample standard deviation
 - n           = sample size
 - df          = degrees of freedom (n - 1)

#  <span style="color:green">16. How is Bayes' Theorem used in statistics, and what is its significance?



</span>

Bayes’ Theorem is a fundamental concept in statistics that describes how to update the probability of a hypothesis based on new evidence.

It provides a mathematical way to revise existing beliefs (prior probabilities) when given new data (evidence), making it essential in fields like machine learning, diagnostics, decision-making, and risk analysis.

🔹 Bayes' Theorem Formula:


                    P(A|B) = [P(B|A) * P(A)] / P(B)

 Where:
-  P(A|B) = Posterior probability (probability of A given B)
- P(B|A) = Likelihood (probability of B given A is true)
- P(A)   = Prior probability of A
- P(B)   = Marginal probability of B


                                                🔹 Significance of Bayes' Theorem:
| Application                         | Example Use                                                  |
|-------------------------------------|---------------------------------------------------------------|
| Medical Diagnosis                   | Updating disease probability after test results              |
| Spam Filtering                      | Determining if an email is spam based on word patterns       |
| Machine Learning                    | Probabilistic classifiers (e.g., Naive Bayes)                |
| Risk Management                     | Updating risks with new information                          |
| Legal & Forensics                   | Revising likelihood of guilt based on new evidence           |

#  <span style="color:green">17. What is the Chi-square distribution, and when is it used?

</span>

The Chi-square (χ²) distribution is a continuous probability distribution that arises in statistics when summing the squares of independent standard normal variables.

It is used primarily in hypothesis testing, especially when dealing with categorical data.



🔹 Key Characteristics:

- Only defined for non-negative values (χ² ≥ 0)
- Shape depends on degrees of freedom (df)

- As df increases, the distribution becomes more symmetric and approaches a normal distribution



✅ Chi-square Formula (for testing):

                      χ² = Σ [ (Oᵢ - Eᵢ)² / Eᵢ ]

 Where:
- Oᵢ = Observed frequency
- Eᵢ = Expected frequency
- Σ  = Summation over all categories



                                            🔹 When is Chi-square Distribution Used?
| Use Case                        | Description                                                  |
|---------------------------------|--------------------------------------------------------------|
| **Goodness-of-fit test**        | Does observed data fit a specific distribution?              |
| **Test for independence**       | Are two categorical variables independent? (e.g., gender vs choice) |
| **Test for homogeneity**        | Are proportions equal across different populations?          |
| **Confidence intervals**        | For population variance and standard deviation               |

#  <span style="color:green">19. What is the F-distribution, and when is it used in hypothesis testing?


</span>

The F-distribution is a continuous probability distribution that arises when comparing two variances.
It is defined as the ratio of two scaled chi-square distributions and is commonly used in hypothesis testing, especially when evaluating models or comparing variability.

🔹 Formula Basis:
 
 If S1² & S2² are sample variances from two independent normal populations:
                        
                         F = (S1² / σ1²) / (S2² / σ2²)

 In practice (especially in F-tests):   

                        F = S1²  / S2² 


Where:

 -  S1² & S2² are sample variances

 - The numerator and denominator must be from independent samples


  - df1 & df2 are the respective degrees of freedom



                                                ✅ When is the F-distribution used?
| Use Case                      | Description                                                 |
|-------------------------------|-------------------------------------------------------------|
| **F-test for equal variances** | Tests whether two populations have the same variance       |
| **ANOVA (Analysis of Variance)** | Tests whether means of three or more groups are equal     |
| **Regression Analysis**        | Tests the significance of a regression model               |

🔹 Characteristics of the F-distribution:

- Only positive values (starts at 0, skewed right)

- Shape depends on two parameters: df1 and df2
 
 

- Approaches normality as degrees of freedom increase

#  <span style="color:green">20. What is an ANOVA test, and what are its assumptions?


</span>

ANOVA (Analysis of Variance) is a statistical method used to determine whether there are significant differences between the means of three or more independent groups.

It helps test if at least one group mean is statistically different from the others, based on the variation within and between groups.

✅ Hypotheses in One-Way ANOVA:

- Null Hypothesis (H₀): All group means are equal
                               𝜇1=𝜇2=𝜇3=⋯=𝜇𝑘

 
- Alternative Hypothesis (H₁): At least one group mean is different


🔹 ANOVA F-statistic Formula:

                   F= Within-group variance (MSW)/Between-group variance (MSB)

    Where:

- MSB = Mean Square Between = 𝑆𝑆𝐵/dfB


- MSW = Mean Square Within = 𝑆𝑆𝑊/𝑑𝑓𝑊

 
- SSB = Sum of Squares Between

- SSW = Sum of Squares Within



✅ Assumptions of ANOVA:

- Independence: Samples are independent from each other

- Normality: Data in each group is approximately normally distributed

- Homogeneity of variances: All groups have equal population variances (can be tested with Levene’s Test or Bartlett’s Test)



#  <span style="color:green">21. What are the different types of ANOVA tests?


</span>

There are several types of ANOVA (Analysis of Variance) tests, each designed for different experimental designs and types of data.
All ANOVA types analyze differences between group means, but they differ based on the number of factors and whether the same subjects are used repeatedly.



✅ 1. One-Way ANOVA

- Purpose: Compares means of 3 or more independent groups based on one independent variable (factor)

- Example: Test scores across 3 different teaching methods


✅ 2. Two-Way ANOVA

- Purpose: Compares means based on two independent variables (factors)

- Can assess main effects and interaction effects

- Example: Effect of teaching method and gender on test scores

🔹 Can be:

- With replication (multiple observations per combination of factors)

- Without replication (only one observation per combination)

✅ 3. Repeated Measures ANOVA

- Purpose: Used when same subjects are measured multiple times under different conditions

- Controls for individual differences

- Example: Measuring blood pressure before, during, and after a treatment on same patients

✅ 4. MANOVA (Multivariate ANOVA)

- Purpose: Extension of ANOVA when there are multiple dependent variables

- Example: Analyzing effect of teaching method on both test score and time taken

| ANOVA Type             | Factors | Repeated Measures? | Use Case                                 |

| One-Way ANOVA          | 1       | No                  | Compare group means by one factor         |

| MANOVA                 | 1+      | Varies              | Multiple dependent variables              |
